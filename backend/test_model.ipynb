```python
import os
import json
import time
from openai import OpenAI

# ==========================
# Khởi tạo client
# ==========================
token = os.getenv("GITHUB_TOKEN")

if not token:
    raise ValueError("Chưa thiết lập GITHUB_TOKEN")

client = OpenAI(
    base_url="https://models.github.ai/inference",
    api_key=token,
)

# ==========================
# Danh sách 20 câu hỏi
# ==========================
queries = [
    "Hãy giới thiệu về Bà Nà Hills.",
    "Có gì thú vị tại Phố cổ Hội An?",
    "Những địa điểm nổi tiếng ở Hà Nội là gì?",
    "Giới thiệu về Vịnh Hạ Long.",
    "Nên tham quan Huế trong bao nhiêu ngày?",
    "Các món ăn đặc sản của Đà Nẵng là gì?",
    "Đến Đà Lạt nên tham quan những đâu?",
    "Giới thiệu về chợ nổi Cái Răng.",
    "Có gì đặc biệt ở đảo Phú Quốc?",
    "Những địa điểm du lịch nổi tiếng ở Ninh Bình?",
    "Giới thiệu về Hang Sơn Đoòng.",
    "Đi Sa Pa mùa nào đẹp nhất?",
    "Các món ăn nổi tiếng của Huế.",
    "Nên đi Mũi Né vào thời điểm nào?",
    "Giới thiệu về Cầu Vàng Đà Nẵng.",
    "Những hoạt động không nên bỏ lỡ tại Phú Quốc.",
    "Có gì thú vị ở Côn Đảo?",
    "Đặc sản nổi tiếng của Hội An.",
    "Giới thiệu về Văn Miếu Quốc Tử Giám.",
    "Những điểm tham quan nổi bật ở Thành phố Hồ Chí Minh."
]

# ==========================
# Sinh dữ liệu
# ==========================
results = []

for idx, question in enumerate(queries, start=1):
    print(f"[{idx}/{len(queries)}] {question}")

    try:
        response = client.chat.completions.create(
            model="openai/gpt-4o-mini",
            temperature=0.2,
            messages=[
                {
                    "role": "system",
                    "content": "Bạn là chuyên gia du lịch Việt Nam. Hãy trả lời ngắn gọn, chính xác."
                },
                {
                    "role": "user",
                    "content": question
                }
            ]
        )

        answer = response.choices[0].message.content

    except Exception as e:
        answer = f"ERROR: {str(e)}"

    results.append({
        "id": idx,
        "question": question,
        "answer": answer
    })

    time.sleep(1)   # tránh gửi request quá nhanh

# ==========================
# Lưu JSON
# ==========================
with open("tourism_qa.json", "w", encoding="utf-8") as f:
    json.dump(results, f, ensure_ascii=False, indent=4)

print("Đã lưu vào tourism_qa.json")
```
